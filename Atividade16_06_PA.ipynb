```python
#QUESTÃO 02 – Soma dos Números (2 pontos)
#---------------------------------------------------------

# Solicite ao usuário um número inteiro positivo.Utilizando um laço de repetição, calcule e exiba a soma detodos os números de 1 até o número informado.

num = int(input("Digite um número: "))
soma = 0

for i in range(num+1):
  soma += i

print(soma)
```

```python
#QUESTÃO 04 – Fatorial (4 pontos)
#---------------------------------------------------------

#Solicite ao usuário um número inteiro positivo e calcule seu fatorial utilizando um laço de repetição.

num = int(input("Digite um número: "))
while num <= 0:
  num = int(input("Digite um número válido: "))

for i in range(1,num):
  num *= i

print(num)
```

```python

#QUESTÃO 05 – Caixa Eletrônico (5 pontos)
#---------------------------------------------------------

#Um caixa eletrônico possui notas de:

#R$ 100
#R$ 50
#R$ 20
#R$ 10
#R$ 5
#R$ 2
#R$ 1

#Solicite ao usuário o valor do saque e informe a menor quantidade possível de notas para realizar o pagamento.

a =int(input("Digite o valor do saque: "))
cem = 0
cinq = 0
vin = 0
dez = 0
cin = 0
dois = 0
um = 0

while a > 0:
  if a >= 100:
    a -= 100
    cem += 1
  elif a >= 50:
    a -= 50
    cinq += 1
  elif a >= 20:
    a -= 20
    vin += 1
  elif a >= 10:
    a -= 10
    dez += 1
  elif a >= 5:
    a -= 5
    cin += 1
  elif a >= 2:
    a -= 2
    dois += 1
  elif a >= 1:
    a -= 1
    um += 1

if cem > 0:
  print(f"{cem}  nota(s) de R$100")
if cinq > 0:
  print(f"{cinq}  nota(s) de R$50")
if vin > 0:
  print(f"{vin}  nota(s) de R$20")
if dez > 0:
  print(f"{dez}  nota(s) de R$10")
if cin > 0:
  print(f"{cin}  nota(s) de R$5")
if dois > 0:
  print(f"{dois}  nota(s) de R$2")
if um > 0:
  print(f"{um}  moeda(s) de R$1")





```
