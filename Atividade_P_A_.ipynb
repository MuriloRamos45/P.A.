```python
#Exercício 2

a = float(input("Digite um número: "))
b = float(input("Digite um número: "))

if a > b:
  print(f"{a} é maior que {b}")
if a < b:
  print(f"{b} é maior que {a}")
if a == b:
  print(f"{a} é igual a {b}")
```

```python
#Exercício 6

a = float(input("Digite um número: "))

if a < 0:
  print(f"{a} é um número negativo")
if a == 0:
  print("Seu número é 0")
if a > 0:
  print(f"{a} é um número positivo")
```

```python
#Exercício 7


a = float(input("Digite o primeiro número: "))
b = float(input("Digite o segundo número: "))
c = str(input("Digite a operação: "))

if c == "+" :
  print(f"O resultado da soma de {a} mais {b} é igual a {a + b}")
elif c == "-" :
  print(f"O resultado da subtração de {a} menos {b} é igual a {a - b}")
elif c == "*" :
  print(f"O resultado da multiplicação de {a} vezes {b} é igual a {a * b}")
elif c == "/" :
  print(f"O resultado da divisão de {a} dividido {b} é igual a {a / b}")
```

```python
#Exercício 10

a = float(input("Digite o valor da compra: "))

if a >= 100:
  totala = a - (a / 10)
  print(f"O valor final da compra com 10% de desconto foi de {totala} reais")
elif a >= 50:
  totalb = a - (a / 20)
  print(f"O valor final da compra com 5% de desconto foi de {totalb} reais")
else:
  print(f"a compra não teve desconto, o valor total foi de {a} reais")

```
