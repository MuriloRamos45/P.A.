```python
#2

a = int(input("Digite um número: "))

if a % 2 == 0:
  print("O número é par")

else:
  print("O número é ímpar")
```

```python
#3

a = int(input("Digite um número: "))
b = int(input("Digite outro número: "))

if a > b:
  print(f"{a} é maior que {b}")
elif b > a:
  print(f"{b} é maior que {a}")
else:
  print("Os números são iguais")
```

```python
#4

a = float(input("Digite a nota do aluno: "))

if a >= 7:
  print("O aluno foi aprovado")
elif a >= 5:
  print("O aluno está de recuperação")
else:
  print("O aluno foi reprovado")
```

```python
#5

a = float(input("Digite o valor da compra: "))

if a >= 100:
  print(f"O valor final da compra é de {a - (a * 0.1)} reais")
else:
  print(f"O valor final da compra é de {a} reais")
```

```python
#6

a = float(input("Digite um número: "))

if a > 0:
  print(f"O número {a} é positivo")
elif a < 0:
  print(f"O número {a} é negativo")
else:
  print("O número é zero")
```

```python
#8

a = float(input("Digite sua nota: "))
b = float(input("Digite sua outra nota: "))

media = (a + b) / 2

if media >= 7:
  print(f"O aluno foi aprovado com a média {media}")
else:
  print(f"O aluno foi reprovado com a média {media}")
```

```python
#10

p = int(input("Digite seu peso em quilos: "))
a = float(input("Digite sua altura em metros: "))

i = p / (a * a)

print(f"Seu IMC é {i}")

if i < 18.5:
  print("Você está abaixo do peso")
elif i < 25:
  print("Você está com o peso normal")
elif i > 30:
  print("Você está sobrepeso")
else:
  print("Você está obeso")
```
