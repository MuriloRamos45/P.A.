```python
#1 Contagem simples, o usuário digita um número e o programa imprimi do zero até o número

num = int(input("Digite um número "))

for i in range (num+1):
  print(i)
```

```python
#2 Contagem Regressiva, o usuário digita um número e o preograma imprimi do número até o zero

num = int(input("Digite um número: "))

for i in range(0 , num+1):
  print(num - i)

```

```python
#3 Números pares. o usuário digita um número e o programa imprimi os números pares

num = int(input("Digite um número: "))

for i in range (num+1):
  if i % 2 == 0:
    print(i)
```

```python
#4 Números impares. o usuário digita um número e o programa imprimi os números impares

num = int(input("Digite um número: "))

for i in range (num+1):
  if i % 2 == 1:
    print(i)
```

```python
#5 Soma, o usuário digita um número e o programa soma os números do zero até o número

num = int(input("Digite um número: "))
soma = 0

for i in range(num+1):
  soma += i

print(soma)
```

```python
#6 Tabuada, o usuário digita um número e o programa imprimi a tabuada do número

num = int(input("Digite um número:"))

for i in range(11):
  print (f"{num} X {i} = {num * i}")
```

```python
#7 Contar letras, o usuário digita uma palavra e o programa conta as letras da palavra

pal = str(input("Digite uma palavra: "))
letras = 0

for i in (pal):
  letras += 1
print(letras)
```

```python
#8 Contar vogais, o usuário digita uma palavra e o programa conta as vogais
palavra = input("Digite uma palavra: ")
vogais = "aeiou"
vogal = 0

for letra in palavra:
    if letra in vogais:
        vogal += 1

print(f"A palavra {palavra} tem {vogal} vogais.")
```

```python
# 9 maior numero digitado, o usuario digita quatro numeros e o programa imprime o maior numero digitado

print("Por favor, digite quatro números.")

num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))
num3 = float(input("Digite o terceiro número: "))
num4 = float(input("Digite o quarto número: "))

numeros = [num1, num2, num3, num4]

maior_numero = numeros[0]

for numero in numeros:
    if numero > maior_numero:
        maior_numero = numero

print(f"O maior número digitado é: {maior_numero}")
```

```python

# 10 O usuaria digita um numero entre 01 e 100 e o programa mostra quais sao divisiveis por 3 e 5

numero_limite = int(input("Digite um número entre 1 e 100: "))

if 1 <= numero_limite <= 100:
    print(f"Números divisíveis por 3 e 5 até {numero_limite}:")
    for i in range(1, numero_limite + 1):
        if i % 3 == 0 and i % 5 == 0:
            print(i)
else:
    print("Por favor, digite um número válido entre 1 e 100.")
```
