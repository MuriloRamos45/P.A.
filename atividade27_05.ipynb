```python
#Jogo menu inteligente

#Criar um sistema que funcione até o usuário escolher sair

#Menu

# 1 - par/impar
# 2 - positivo/negativo
# 3 - maior
# 4 - sair



opcao = 0

while opcao != 4:
  print("--- Bem Vindo! ---")
  print("1 - Par / Ímpar")
  print("2 - Positivo / Negativo")
  print("3 - Maior")
  print("4 - Sair")
  print("------------")

  opcao = int(input("Escolha uma opção: "))

  if opcao == 1:
      numero = int(input("Digite um número: "))
      if numero % 2 == 0:
          print(f"O número {numero} é par")
      else:
          print(f"O número {numero} é ímpar")
  elif opcao == 2:
      numero = int(input("Digite um número para verificar (positivo/negativo): "))
      if numero > 0:
          print(f"O número {numero} é positivo.")
      elif numero < 0:
          print(f"O número {numero} é negativo.")
      else:
          print(f"O número {numero} é zero.")
  elif opcao == 3:
      num1 = int(input("Digite o primeiro número: "))
      num2 = int(input("Digite o segundo número: "))
      if num1 > num2:
          print(f"O maior número é: {num1}")
      elif num2 > num1:
          print(f"O maior número é: {num2}")
      else:
          print("Ambos os números são iguais.")
  elif opcao == 4:
      print("Saindo do programa. Até mais!")
  else:
      print("Opção inválida. Por favor, escolha um número de 1 a 4.")
```

```python
# Urna Eletrônica

# Criar um programa de votação com 3 candidatos

# Votar candidato A
# Votar candidato B
# Votar candidato C
# Votar Branco
# Votar Nulo
# Encerrar votação

# O sistema deve repetir até encerrar a votação
# Contar votos de cada candidato
# Contar votos brancos e nulos
# Ao encerrar, mostrar:

# Total de Votos
# Total de votos por candidato
# Total votos brancos
# Total votos nulos
# Vencedor da eleição

# Regras extras

# Se houver empate entre candidatos mostrar: eleição empatada
# Se não houver votos válidos, mostrar: "Nenhun candidato recebeu voto"

opcao = 0
votosa = 0
votosb = 0
votosc = 0
votosbranco = 0
votosnulo = 0


print("--- Bem Vindo! ------")
print("1 - Votar candidato A")
print("2 - Votar candidato B")
print("3 - Votar candidato C")
print("4 - Votar branco")
print("5 - Votar nulo")
print("6 - Encerrar votação")
print("---------------------")
print("")
while opcao != 6:

  print("")
  opcao = int(input("Escolha uma opção: "))


  if opcao == 1:
      print("Você votou no Candidato A")
      votosa += 1
  elif opcao == 2:
      print("Você votou no Candidato B")
      votosb += 1
  elif opcao == 3:
      print("Você votou no Candidato C")
      votosc += 1
  elif opcao == 4:
      print("Você votou branco")
      votosbranco += 1
  elif opcao == 5:
      print("Você votou nulo")
      votosnulo += 1
  elif opcao == 6:
      print("Encerrando votação")


  else:
      print("Digite uma opção válida")


print(f"{votosa + votosb + votosc + votosbranco + votosnulo} pessoas participaram da votação")
print(f"O candidato A teve {votosa} votos")
print(f"O candidato B teve {votosb} votos")
print(f"O candidato C teve {votosc} votos")
print(f"{votosbranco} eleitores votaram em branco")
print(f"{votosnulo} eleitores votaram nulo")
print("")
if votosa > votosb and votosa > votosc:
  print("O candidato A ganhou a eleição")
elif votosb > votosa and votosb > votosc:
  print("O candidato B ganhou a eleição")
elif votosc > votosb and votosc > votosa:
  print("O candidato C ganhou a eleição")
elif votosa == 0 and votosb == 0 and votosc == 0:
  print("Não teve votos válidos")
elif votosa == votosb and votosb == votosc:
  print("Houve um empate entre os três candidatos")
elif votosa == votosb:
  print("Houve um empate entre os candidatos A e B")
elif votosa == votosc:
  print("Houve um empate entre os candidatos A e C")
elif votosc == votosb:
  print("Houve um empate entre os candidatos B e C")

```
