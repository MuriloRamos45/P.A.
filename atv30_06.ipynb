```python
#Cofre digital

#Um cofre possui um código composto por 5 números inteiros

#Desenvolva um programa que:

#Leia os 5 números utilizando um laço de repetição

#Ao final informar:

# 1- Quantos números são pares.
# 2- Quantos números são ímpares
# 3- O Maior número
# 4- A Soma de todos os números.
# 5- A média dos números
# 6- O menor número
# 7- Quantos números são maiores que a média

pares = 0
impares = 0
soma = 0

num = int(input("Digite o 1º número: "))
maior = num
menor = num
soma += num

if num % 2 == 0:
    pares += 1
else:
    impares += 1

numeros = [num]

for i in range(2, 6):
    num = int(input(f"Digite o {i}º número: "))
    numeros.append(num)
    soma += num

    if num % 2 == 0:
        pares += 1
    else:
        impares += 1

    if num > maior:
        maior = num

    if num < menor:
        menor = num

media = soma / 5

maiores_media = 0
for num in numeros:
    if num > media:
        maiores_media += 1

print("\nResultado:")
print("1 - Quantidade de números pares:", pares)
print("2 - Quantidade de números ímpares:", impares)
print("3 - Maior número:", maior)
print("4 - Soma dos números:", soma)
print("5 - Média dos números:", media)
print("6 - Menor número:", menor)
print("7 - números maiores que a média:", maiores_media)
```

```python
# Número espelho

# Solicite um número inteiro positivo

# Inverta o número
# Informe se ele é um palindromo


numero = int(input("Digite um número inteiro positivo: "))


numeroinvertido = str(numero)[::-1]


print("Número invertido:", numeroinvertido)


if str(numero) == numeroinvertido:
    print("O número é um palíndromo.")
else:
    print("O número não é um palíndromo.")
```
